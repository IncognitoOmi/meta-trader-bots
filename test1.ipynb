```python
class A:
    def execute(self):
        print('A')

class B(A):
    def execute(self):
        print('B')
        super().execute()

class C(A):
    def execute(self):
        print('C')
        super().execute()

class D(B,C):
    def execute(self):
        print('D')
        super().execute()

d = D()
d.execute()
```

```python
c = [10, 20, 30,100,5,50,500]
pairs = []
for i in range(len(c)):
    pairs.append(min(c))
    for j in range(1,len(c)):
        pairs.append(max(c))

pairs
```
